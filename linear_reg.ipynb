```python
from sklearn.decomposition import PCA
```

```python
p=PCA(n_components=1)
```

```python
df=l.frame
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
xtr,xts,ytr,yts=train_test_split(df.drop(columns=["target"]),df.target)
```

```python
lr=LinearRegression()
lr.fit(xtr,ytr)
lr.score(xts,yts)
```

```python
xts=p.fit_transform(xts)
```

```python
xtr=p.transform(xtr)
```

```python
lr=LinearRegression()
lr.fit(xtr,ytr)
lr.score(xts,yts)
```

```python
xtr.shape
```

```python
class linear :
    def __init__(self):
        self.m=None
        self.c=None
    def solve(self,x,y):
        num=0
        den=0
        xbar=x.mean()
        ybar=y.mean()
        for i in range (x.shape[0]):
            num+=(x[i]-xbar)*(y.iloc[i]-ybar)
            den+=(x[i]-xbar)**2
        self.m=num/den
        self.c=ybar-(self.m)*xbar
    def prt(self,x,y):
        for i in range(x.shape[0]):
            # print(x[i],y.iloc[i])
            return (int(np.round(self.m*x[i]+self.c)))
```

```python
l=linear()
from sklearn.metrics import accuracy_score
```

```python
l.solve(xtr,ytr)
```

```python
l.prt(xts,yts)
```

```python
from sklearn.metrics import accuracy_score

y_pred = np.rint(l.m * xts[:, 0] + l.c)
y_pred = np.clip(y_pred, ytr.min(), ytr.max()).astype(int)

print("Accuracy:", accuracy_score(yts.to_numpy(), y_pred))
```
